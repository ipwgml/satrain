The results are intended to provide a common point of reference for the development and evaluation of new precipitation retrieval algorithms across different sensor types and retrieval techniques.

## Base Sensor: GMI

### Quantitaive Precipitation Estimation


```python
from IPython.display import HTML, display
from satrain.baselines import get_ranked_results, plot_baselines_interactive, QPE_METRICS

results = get_ranked_results(base_sensor="gmi", metrics=list(QPE_METRICS))
fig = plot_baselines_interactive(results)
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```

### Precipitation Detection

The precipitation-detection skill of all retrievals is evaluated by testing their ability to identify reference precipitation rates exceeding 0.1 mm h$^{-1}$. If an algorithm provides a precipitation flag or mask, the mask is used to calculate FAR, POD, and HSS. If the algorithm additionally provides a probability of precipitation, it is evaluated using the area under the Precision-Recall (PR) curve. If an algorithm only provide dedicated outputs for the detection of precipitation, the areas where the precipitation rate exceeds 0.1 mm h$^{-1}$ are considered to be raining.

```python
from satrain.baselines import DETECTION_METRICS

results = get_ranked_results(base_sensor="atms", metrics=list(DETECTION_METRICS))
fig = plot_baselines_interactive(results)
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```

### Precipitation Detection

The heavy-precipitation-detection skill of all retrievals is evaluated by testing their ability to identify reference precipitation rates exceeding 10.0 mm h$^{-1}$. If an algorithm provides a precipitation flag or mask, the mask is used to calculate FAR, POD, and HSS. If the algorithm additionally provides a probability of heavy precipitation, it is evaluated using the area under the Precision-Recall (PR) curve. If an algorithm only provides a precipitation rat, the areas where the precipitation rate exceeds 10.0 mm h$^{-1}$ are considered to be raining.

```python
from satrain.baselines import HEAVY_DETECTION_METRICS
results = get_ranked_results(base_sensor="atms", metrics=list(HEAVY_DETECTION_METRICS))
fig = plot_baselines_interactive(results)
display(HTML(fig.to_html(full_html=False, include_plotlyjs="cdn")))
```
